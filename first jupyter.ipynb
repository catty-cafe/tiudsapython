# Python Basics: Introduction to Jupyter Notebooks
This notebook demostrates foundational Python programming concepts,including print statements, comments,variables, basic arithmetic, input/output operations, and f-string formatting.

```python
# This is a single-line comment. The code below uses print statements to output text.
print("Hello, World!")
print("Welcome to learning Python basics in Jupyter Notebooks.")

"""
This is a multi-line comment (or docstring). 
It can span multiple lines to explain larger blocks of code or logic.
"""
print("Comments are ignored by the Python interpreter.")
```

```python
# Defining variables of different data types
item_name = "Notebook"
quantity_in_stock = 15
unit_price = 3.50
is_available = True

# Basic arithmetic operations
total_inventory_value = quantity_in_stock * unit_price
subtotal = 50 + 25
remainder = 10 % 3  # Modulus operator

print(f"Item: {item_name}")
print(f"Total Inventory Value: ${total_inventory_value}")
print(f"Subtotal: {subtotal}")
print(f"Remainder of 10 / 3: {remainder}")
```

```python
## Taking user input (Note: input() returns a string)
user_name = input("Please enter your name: ")

print(f"Hello, {user_name}! It is great to meet you.")

# Converting input types for arithmetic
user_age = input("Please enter your age: ")
age_next_year = int(user_age) + 1

print(f"Next year, you will be {age_next_year} years old.")
```

```python
# Formatting numbers and floating-point values using f-strings
pi_approx = 3.1415926535
large_number = 1000000

print(f"Pi rounded to two decimal places: {pi_approx:.2f}")
print(f"Formatted large number with commas: {large_number:,}")
```
